## Plot 1

**Scenario**: A researcher wants to show that Gentoo penguins are significantly heavier than other species, but their plot is visually overwhelming and exaggerates the differences.

```python
penguins = sns.load_dataset("penguins")

ax = sns.barplot(data=penguins, x="species", y="body_mass_g", color="steelblue")

ax.set_ylim(0, 5500)
ax.set_xlabel("Species")
ax.set_ylabel("Average Body Mass (g)")
plt.title("Average Penguin Body Mass by Species")

plt.show()
```

Issues with the plot: The y-axis starts at 3000 instead of 0, so the difference in body mass looks much bigger than it really is. The bright colors are also unnecessary because they don't represent any extra information.

How you changed the plot and why: I changed the y-axis to start at 0 so the comparison is more accurate. I also used one simple color and added clearer labels to make the graph easier to read.

## Plot 2

**Scenario**: A student plotted the carat size versus price for 54,000 diamonds to see where the bulk of the market sits, but the result is just a solid block of ink.

```python
diamonds = sns.load_dataset("diamonds")

sns.scatterplot(
    data=diamonds,
    x="carat",
    y="price",
    color="blue",
    s=10,
    alpha=0.2
)

plt.xlabel("Carat")
plt.ylabel("Price ($)")
plt.title("Diamond Price vs. Carat")
plt.show()
```

Issues with the plot: There are too many points and the markers are too large, so a lot of them overlap. This makes it hard to see where most of the diamonds are concentrated.

How you changed the plot and why: I made the points smaller and added transparency. This makes the graph less crowded and makes areas with more diamonds easier to see.

## Plot 3

The aviation board wants to see passenger growth month-over-month for each year in the 1950s, but the lines are indistinguishable.

```python
flights = sns.load_dataset("flights")

flights_1950s = flights[
    (flights["year"] >= 1950) & (flights["year"] <= 1959)
].copy()

flights_1950s["year"] = flights_1950s["year"].astype(str)

sns.lineplot(
    data=flights_1950s,
    x="month",
    y="passengers",
    hue="year",
    palette="viridis"
)

plt.xlabel("Month")
plt.ylabel("Passengers")
plt.title("Monthly Airline Passengers by Year")
plt.legend(title="Year")
plt.show()
```

Issues with the plot: All of the years use the same black color, so it is hard to tell which line belongs to each year. There is also no legend showing the different years.

How you changed the plot and why: I used different colors for each year and added a legend. I focused only on the years from 1950 to 1959, which match the scenario and makes the graph easier to compare.

## Plot 4

**Scenario**: A restaurant manager receives this chart showing the relationship between total bills and tips, split by smoking preference. 





```python
tips = sns.load_dataset("tips")

g = sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    ci=None,
    scatter_kws={"alpha": 0.6, "s": 40}
)

g.set_axis_labels("Total Bill ($)", "Tip ($)")
g.fig.suptitle("Total Bill vs. Tip by Smoking Preference", y=1.02)

plt.show()
```

Issues with the plot: The axis labels use the original column names, which are not very clear. The confidence areas also make the plot look more crowded, and some of the points overlap.

How you changed the plot and why: I removed the confidence areas and made the points slightly transparent. I also added clearer axis labels and a title so the graph is easier to understand.
